```python
import numpy as np

# Nhiệm vụ 1: Đọc CSV..........................................
dataA = np.genfromtxt("campaign_A.csv", delimiter=",", dtype=str, skip_header=1)
dataB = np.genfromtxt("campaign_B.csv", delimiter=",", dtype=str, skip_header=1)

# Nhiệm vụ 2: Tách cột............................................
dateA = dataA[:,0]
visitsA = dataA[:,1].astype(int)

dateB = dataB[:,0]
visitsB = dataB[:,1].astype(int)

# Nhiệm vụ 3: Chuyển cột ngày.......................................
date_changeA = dateA.astype('datetime64[D]')
date_changeB = dateB.astype('datetime64[D]')

# Nhiệm vụ 4: Kiểm tra dtype và shape................................................
print(f"Kieu du lieu: {date_changeA.dtype} - {date_changeB.dtype}")
print(f"Shape A: {date_changeA.shape} - Shape B: {date_changeB.shape}")

# In 5 giá trị đầu/cuối
print("5 giá trị đầu campaignA:", visitsA[:5], "5 giá trị cuối:", visitsA[-5:])
print("5 giá trị đầu campaignB:", visitsB[:5], "5 giá trị cuối:", visitsB[-5:])

# Nhiệm vụ 5: Slicing.............................................
# 7 ngày đầu
print("7 ngày đầu:")
for d, v in zip(date_changeA[:7], visitsA[:7]): # dùng zip để truy cập song song các phần tử của hai mảng dates và visits
    print(d, v) # in date và visit tương ứng 

# 7 ngày giữa
print("\n7 ngày giữa:")
for d, v in zip(date_changeA[11:18], visitsA[11:18]):
    print(d, v) 

# 7 ngày cuối
print("\n7 ngày cuối:")
for d, v in zip(date_changeA[-7:], visitsA[-7:]):
    print(d, v)  

# Nhiệm vụ 6: Boolean mask A > B...........................................
mask = visitsA > visitsB
print("Ngày A > B:", date_changeA[mask])

# Nhiệm vụ 7: Mask tăng đột biến B..............................................
mask = visitsB > 2000
print("Ngày B tăng đột biến:", date_changeB[mask])

# Nhiệm vụ 8: Fancy Indexing................................................
print("Top 5 ngày truy cập cao nhất A:")
top5 = np.argsort(visitsA)[::-1][:5]
print(date_changeA[top5])

print("Top 5 ngày truy cập cao nhất B:")
top_5 = np.argsort(visitsB)[::-1][:5]
print(date_changeB[top_5])

# Nhiệm vụ 9: Broadcasting............................................
chenh_lech = visitsA - visitsB
print("Chênh lệch A - B:", chenh_lech)

# Nhiệm vụ 10: np.diff...................................................
print("Biến động A:", np.diff(visitsA))
print("Biến động B:", np.diff(visitsB))

# Nhiệm vụ 11: Thống kê......................................................
print("Thống kê A:")
print(visitsA.mean(), visitsA.std(), visitsA.max(), visitsA.min(), visitsA.argmax(), visitsA.argmin())

print("Thống kê B:")
print(visitsB.mean(), visitsB.std(), visitsB.max(), visitsB.min(), visitsB.argmax(), visitsB.argmin())

# Nhiệm vụ 12: Tạo bảng tổng hợp...................................................
tao_bang = np.array([
    ["Campaign", "A", "B"],
    ["Mean", visitsA.mean(), visitsB.mean()],
    ["Std", visitsA.std(), visitsB.std()],
    ["Max", visitsA.max(), visitsB.max()],
    ["Min", visitsA.min(), visitsB.min()],
    ["Argmax", visitsA.argmax(), visitsB.argmax()],
    ["Argmin", visitsA.argmin(), visitsB.argmin()],
])

print(tao_bang)

# Lưu file
np.savetxt("luu_ket_qua_case_3.csv", tao_bang, fmt="%s", delimiter=",")

print("Đã lưu thành công")









```
